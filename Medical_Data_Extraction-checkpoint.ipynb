```python
#Dataset load karo
import pandas as pd
import os

file_path = r"C:\Users\nishika tendane\Downloads\archive (9)\mtsamples.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
#Dataset check karo
print("Column Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
#: Medical text ko clean karo
import re

# Check the transcription column
print(df["transcription"].head())
```

```python
def clean_text(text):
    if pd.isna(text):
        return ""

    text = str(text)
    text = re.sub(r"\s+", " ", text)
    text = text.strip()

    return text

df["cleaned_transcription"] = df["transcription"].apply(clean_text)

df[["transcription", "cleaned_transcription"]].head()
```

```python
#Cleaned dataset check karo
print("Total records:", len(df))
print("Empty medical notes:", (df["cleaned_transcription"] == "").sum())

df[["medical_specialty", "cleaned_transcription"]].head()
```

```python
#Medical text se information extract karna
import re
import pandas as pd

def extract_medical_info(text):
    if pd.isna(text):
        text = ""
    else:
        text = str(text)

    # Extract age
    age = re.search(r'\b(\d{1,3})[- ]year[- ]old\b', text, re.I)

    # Extract common medical terms
    patterns = {
        "diagnosis": r'\b(?:diagnosis|impression|assessment)\s*:\s*([^.;\n]+)',
        "medication": r'\b(?:medication|medications|prescribed|prescription)\s*:\s*([^.;\n]+)',
        "procedure": r'\b(?:procedure|procedures|operation)\s*:\s*([^.;\n]+)'
    }

    result = {
        "extracted_age": age.group(1) if age else None
    }

    for field, pattern in patterns.items():
        match = re.search(pattern, text, re.I)
        result[field] = match.group(1).strip() if match else None

    return pd.Series(result)


extracted_info = df["cleaned_transcription"].apply(extract_medical_info)

df = pd.concat([df, extracted_info], axis=1)

df[["medical_specialty", "extracted_age",
    "diagnosis", "medication", "procedure"]].head(10)
```

```python
#Extraction results check karo
print("Extracted age:", df["extracted_age"].notna().sum())
print("Extracted diagnosis:", df["diagnosis"].notna().sum())
print("Extracted medication:", df["medication"].notna().sum())
print("Extracted procedure:", df["procedure"].notna().sum())
```
